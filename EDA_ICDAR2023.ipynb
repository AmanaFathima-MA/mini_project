```python
import json
import pandas as pd

# 1. Load your existing FIR_details.json
json_path = "fir_dataset_icdar2023/FIR_details.json"

with open(json_path, "r", encoding="utf-8") as f:
    raw_data = json.load(f)

df = pd.DataFrame(raw_data)

category_map = {
    0: "Police Station",
    1: "Year",
    2: "Statutes",
    3: "Complainant Name"
}
df['category_name'] = df['category_id'].map(category_map)

# 2. Extract structured fields per image
icdar_samples = []

grouped = df.groupby('image_name')

for idx, (image_name, group) in enumerate(grouped):
    # Extract fields safely
    ps_series = group[group['category_id'] == 0]['text']
    year_series = group[group['category_id'] == 1]['text']
    statute_series = group[group['category_id'] == 2]['text']
    name_series = group[group['category_id'] == 3]['text']

    ps = ps_series.iloc[0].strip() if not ps_series.empty else "Local PS"
    year = year_series.iloc[0].strip() if not year_series.empty else "2023"
    statute = statute_series.iloc[0].strip() if not statute_series.empty else "Relevant Sections"
    complainant = name_series.iloc[0].strip() if not name_series.empty else "Anonymous Complainant"

    # Construct narrative & summary from ICDAR real data
    fir_id = f"FIR/ICDAR/{year}/{idx+1:03d}"
    
    full_text = (
        f"On formal complaint received at {ps} Police Station in the year {year}, "
        f"complainant {complainant} reported an offense involving violation of {statute}. "
        f"The incident occurred under the jurisdiction of {ps} PS. "
        f"Following preliminary investigation, a formal FIR was registered under {statute} for further legal proceedings."
    )
    
    summary = (
        f"Complainant {complainant} filed a complaint at {ps} Police Station in {year}. "
        f"The case was registered under {statute} for legal investigation."
    )

    sample = {
        "fir_id": fir_id,
        "crime_category": statute,
        "full_text": full_text,
        "summary": summary,
        "source": "ICDAR_2023",
        "image_name": image_name
    }
    icdar_samples.append(sample)

# Save converted ICDAR samples to JSON
with open("icdar_samples.json", "w", encoding="utf-8") as f:
    json.dump(icdar_samples, f, indent=2, ensure_ascii=False)

print(f"✅ Successfully converted {len(icdar_samples)} ICDAR FIR records into summarization samples!")
print(f"Saved to: icdar_samples.json\n")

# Preview first sample
print("Example Sample Generated:")
print(json.dumps(icdar_samples[0], indent=2))
```

```python
import json

# 1. Load both JSON files from your folder
with open("llm_samples.json", "r", encoding="utf-8") as f:
    llm_data = json.load(f)

with open("icdar_samples.json", "r", encoding="utf-8") as f:
    icdar_data = json.load(f)

# 2. Combine the lists
master_data = llm_data + icdar_data

# 3. Save to master_fir_dataset.json
with open("master_fir_dataset.json", "w", encoding="utf-8") as f:
    json.dump(master_data, f, indent=2, ensure_ascii=False)

print(f"🎉 Success! Combined {len(llm_data)} LLM + {len(icdar_data)} ICDAR samples.")
print(f"Total Master Dataset Size: {len(master_data)} FIR records saved to 'master_fir_dataset.json'!")
```
